```python
import pandas as pd
import requests

# 1. Configuration des IDs Google Drive
FILE_ID_ADS = '1_8quOdA863-70Q-vOfIRrvC6qe9eGAht'
# Remplacez ceci par l'ID exact du fichier XML/CSV présent dans le dossier du flux
FILE_ID_FLUX = 'VOTRE_FILE_ID_DU_FICHIER_FLUX'

# 2. Fonction de téléchargement générique pour Google Drive
def download_drive_file(file_id, output_filename):
    url = f'https://drive.google.com/uc?id={file_id}&export=download'
    response = requests.get(url)
    with open(output_filename, 'wb') as f:
        f.write(response.content)
    print(f"Fichier {output_filename} téléchargé avec succès.")

# 3. Téléchargement et chargement du rapport Google Ads (.xlsx)
download_drive_file(FILE_ID_ADS, 'ads_data.xlsx')

# Charger les données Ads (Note: Ajuster 'skiprows' si Google Ads a inclus des lignes d'en-tête de rapport)
df_ads = pd.read_excel('ads_data.xlsx')
print(f"Aperçu des données Ads ({len(df_ads)} lignes) :")
display(df_ads.head())

# 4. Téléchargement et chargement du flux produit
# (Exemple si le fichier est un XML)
download_drive_file(FILE_ID_FLUX, 'flux_shopping.xml')
df_flux = pd.read_xml('flux_shopping.xml')
print(f"Aperçu des données Flux ({len(df_flux)} produits) :")
display(df_flux.head())
```
